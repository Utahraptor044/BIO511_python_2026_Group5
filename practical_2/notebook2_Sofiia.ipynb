### Nested loops

Create a nested for loop as in the example which looks for start and stop codons in the sequences list from the example above. Which sequences have both a start and a stop codon?
Hint: Start codons are ATG and stop codons are TAA, TAG and TGA.

```python
sequences = ['1ATCTGAGTCCACACATG', '2GCGTCGTGCGATGTTCACGTTGAT', '3CAGTAGTACTCAGT', '4GGTATGCTAGACGAGATCTAATA']
```

```python
start_codon = "ATG"
stop_codons = ['TAA', 'TAG', 'TGA']
```

```python
for sequence in sequences:
  if start_codon in sequence:
    print("Start codon", start_codon + " is in " + sequence)
  for stop_codon in stop_codons:
    if stop_codon in sequence:
      print("Stop codon", stop_codon + " is in " + sequence)
```

```python
for sequence in sequences:
    if start_codon in sequence:
        start_position = len(sequence.split(start_codon)[0]) + 1
        print("Start codon", start_codon + " is in " + sequence + " at position", start_position)
    for stop_codon in stop_codons:
      if stop_codon in sequence:
        stop_position = len(sequence.split(stop_codon)[0]) + 1
        if start_position<stop_position:
            print("Stop codon", stop_codon + " is in " + sequence + " at position", stop_position)
  
```

```python
sequences_with_start_and_end = []
```

```python
for sequence in sequences:
    for stop_codon in stop_codons:
        if start_codon and stop_codon in sequence:
            print("Sequence ", sequence, "has start codon", start_codon, "and stop codon", stop_codon)
            sequences_with_start_and_end.append(sequence)
print("Sequences with both a start and stop codons:", set(sequences_with_start_and_end))
```

```python
for sequence in sequences:
    for stop_codon in stop_codons:
        if start_codon and stop_codon in sequence:
            start_position = len(sequence.split(start_codon)[0]) + 1
            stop_position = len(sequence.split(stop_codon)[0]) + 1
            if start_position<stop_position:
                print(sequence, start_position, stop_position)
```

```python
for sequence in set(sequences_with_start_and_end):
    for stop_codon in stop_codons:
        if start_codon and stop_codon in sequence:
            start_position = len(sequence.split(start_codon)[0]) + 1
            stop_position = len(sequence.split(stop_codon)[0]) + 1
            if start_position < stop_position:
                print("Sequence", sequence, "has start codon before stop codon at", start_position, "and stop codon at", stop_position)
```
